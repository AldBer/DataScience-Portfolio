```python
import fiona
import geopandas as gpd

shp_file = r"f:\Documents\04. Cursos\17.Python\05.Portfolio\dashboards\01_SP_Housing\data\geo\SIRGAS_SHP_distrito.shp"

# Listar camadas (normalmente apenas 1 para shapefiles simples)
print(fiona.listlayers(shp_file))

# Ver metadados completos
with fiona.open(shp_file) as src:
    print("\nCRS:", src.crs)  # Sistema de coordenadas
    print("Driver:", src.driver)  # Tipo de arquivo
    print("Número de feições:", len(src))  # Quantidade de elementos

shp_file = r"f:\Documents\04. Cursos\17.Python\05.Portfolio\dashboards\01_SP_Housing\data\geo\SIRGAS_SHP_distrito.shp"
distritos = gpd.read_file(shp_file)

# Defina o CRS como SIRGAS 2000 / UTM 23S (EPSG:31983)
distritos.crs = "EPSG:31983"

# Verificação
print(f"CRS definido: {distritos.crs}")  # Deve retornar "EPSG:31983"
print(distritos.head())

# Salve corrigido
output_path = r"f:\Documents\04. Cursos\17.Python\05.Portfolio\dashboards\01_SP_Housing\data\processed\sp_distritos_fixed_crs.geojson"
distritos.to_file(output_path, driver="GeoJSON")
```
