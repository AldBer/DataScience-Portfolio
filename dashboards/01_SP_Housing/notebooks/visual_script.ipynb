```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# 1. Carregue seus dados primeiro
df = pd.read_csv(r'F:/Documents/04. Cursos/17.Python/05.Portfolio/dashboards/01_SP_Housing/data/raw/sp_properties_sample.csv')

# 2. Análise de Preços (exemplo)
plt.figure(figsize=(10,6))
df['preco'].plot(kind='hist', bins=50)
plt.title('Distribuição de Preços de Imóveis em SP')
plt.savefig('../assets/images/distribuicao_precos.png')

# 3. Mapa de Calor (requer lat/long)
if all(col in df.columns for col in ['latitude', 'longitude']):
    import folium
    from folium.plugins import HeatMap
    
    mapa = folium.Map(location=[-23.5505, -46.6333], zoom_start=11)
    HeatMap(data=df[['latitude', 'longitude']].dropna().values).add_to(mapa)
    mapa.save('../assets/interactive/mapa_calor.html')
```
